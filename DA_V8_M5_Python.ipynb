```python
df['hashtags'] = df['hashtags'].fillna('')
df['hashtag_count'] = df['hashtags'].apply(
    lambda x: len([tag for tag in x.split() if tag.startswith('#')])
)

# Clean sentiment
df['sentiment'] = df['sentiment'].str.lower().str.strip()
df['sentiment'] = df['sentiment'].replace({
    'pos': 'positive',
    'neg': 'negative',
    'neu': 'neutral'
})
```
